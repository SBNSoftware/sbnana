```python
# FUNCTION TO REPORT SIGNAL AND BACKGROUND IN ABCD

def checkABCD(cut1, signalthresh_1, var_1, xbins, myrange_x, xtitle,
              cut2, signalthresh_2, var_2, ybins, myrange_y, ytitle,
              scale_1 = 1., B_bounds=None,
              scale_2 = 1., C_bounds=None
             ):
    
    # APPLY THE CUTS:
    
    final_cuts = [ not_stub_05, not_stub_1, not_stub_2, not_stub_3,
              max_shw_len_mask,
              ok_chi2mu_contained, #ok_chi2mu,
              ndaughters_per_conTrk,
              numi_angle_hybrid_mcsCorr_mask, open_angle_mask,
              shortTrk_len_mask
             ]

    nm2_cuts = final_cuts
    nm2_cuts.remove(cut1)
    nm2_cuts.remove(cut2)

    nm2_results = apply_cuts(evtdf, nm2_cuts, # sneaky_BG_cats = True)
                             detailed_bsm=True, detailed_nu='int_type', flip_last_cut=False) # detailed_nu='int_type'
    nm2_mask = nm2_results[-1]
    nm2_df = add_hdr_info(evtdf[nm2_mask], hdrs)
    
    # MAKE THE PLOTS:
    
    fig, axs = plt.subplots(nrows=1, ncols=2, figsize=(12, 5))


    cats = make_categories(nm2_df, detailed_bsm=False, detailed_nu=None)
    ptitles = ['Background', 'Signal']
    for a, ax in enumerate(axs):
        if a==0:
            df = nm2_df.drop(index = nm2_df[make_categories(nm2_df)[0]].index) # drop the signal
        else:
            df = nm2_df[cats[0]] # consider only the signal
        hb, xedges, yedges, im = ax.hist2d(df[var_1]*scale_1,
                                           df[var_2]*scale_2,
                                   
                                           bins=[xbins, ybins], #norm=mpl.colors.LogNorm(),
                                           cmin=1, cmap='inferno', range=(myrange_x, myrange_y))
        fig.colorbar(im, ax=ax, label="Slices")#, "log10(Entries)")
        ax.set_xlabel(xtitle)
        ax.set_ylabel(ytitle)
        ax.set_title('n-2 cuts applied \n'+ptitles[a])
        
        axs[a].axvline(x=signalthresh_1, color='r', linestyle='--', linewidth=2, label='Vertical Line')
        axs[a].axhline(y=signalthresh_2, color='r', linestyle='--', linewidth=2, label='Horizontal Line')

        #plt.plot([0,8], [0,0], color='black', label='y=0', linestyle=':')
        #plt.legend(loc='upper left')
    plt.show()
    plt.close()
    
    # MAKE A TABLE FOR THE # OF SIGNAL & BG EVENTS IN EACH REGION:
    
    #if B_bounds is None:
    #    B_bounds = (signalthresh_1, myrange_x[1])
    #if C_bounds is None:
    #    C_bounds = (signalthresh_1, myrange_x[1])
    
    # initialze data frames
    categories = make_categories(nm2_df, detailed_bsm=True, detailed_nu='int_type')
    sb_counts_df = pd.DataFrame(
        np.zeros((0,len(categories)), dtype=int), # start w/ zero rows (cuts), fill later
        columns = [c.name for c in categories]
    )
    
    # fill in first row of data frame for "no cuts"
    rowA = []
    rowB = []
    rowC = []
    rowD = []
    for c in categories:
        c1_mask = cut1(nm2_df)[0]
        #flipped_c1_mask = ~c1_mask
        c2_mask = cut2(nm2_df)[0]
        #flipped_c2_mask = ~c2_mask
        rowA.append(round(100*sum(nm2_df[c & c1_mask & c2_mask].scale*nm2_df[c & c1_mask & c2_mask].wgt.cv.tot))/100.)
        rowB.append(round(100*sum(nm2_df[c & ~c1_mask & c2_mask].scale*nm2_df[c & ~c1_mask & c2_mask].wgt.cv.tot))/100.)
        rowC.append(round(100*sum(nm2_df[c & c1_mask & ~c2_mask].scale*nm2_df[c & c1_mask & ~c2_mask].wgt.cv.tot))/100.)
        rowD.append(round(100*sum(nm2_df[c & ~c1_mask & ~c2_mask].scale*nm2_df[c & ~c1_mask & ~c2_mask].wgt.cv.tot))/100.)
    sb_counts_df.loc["Region_A"] = rowA
    sb_counts_df.loc["Region_B"] = rowB
    sb_counts_df.loc["Region_C"] = rowC
    sb_counts_df.loc["Region_D"] = rowD
    sb_counts_df['Total BG'] = sb_counts_df.sum(axis=1)
    sb_counts_df.loc["A/C"] = sb_counts_df.loc["Region_A"]/sb_counts_df.loc["Region_C"]
    sb_counts_df.loc["B/D"] = sb_counts_df.loc["Region_B"]/sb_counts_df.loc["Region_D"]

    sb_counts_df = sb_counts_df.drop(sb_counts_df.columns[0:17], axis=1)
    display(sb_counts_df)
    
    return sb_counts_df
    
```
